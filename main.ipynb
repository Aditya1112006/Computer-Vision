```python
import cv2
import matplotlib.pyplot as plt
```

### Reading Image

```python
img = cv2.imread('image.png')
plt.imshow(img)
plt.axis('off')
```

# Conversion of image from bgr to different color type

```python
img = cv2.imread('image.png')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
plt.imshow(img_rgb)
plt.axis('off')
```

```python
img = cv2.imread('image.png')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
plt.imshow(img_rgb)
plt.axis('off')
```

```python
img = cv2.imread('image.png')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(img_rgb)
plt.axis('off')
```

### Use of cmap to convert color to grey

```python
img = cv2.imread('image.png')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(img_rgb,cmap="grey")
plt.axis('off')
```

```python
img = cv2.imread('image.png')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.axis('off')
```
